```python
import pandas as pd
import numpy as np
np.random.seed(42)
n=28
base=pd.DataFrame({
    '订单号': range(1001,1001+n),
    '日期': pd.date_range('2024-01-01', periods=n, freq='D'),
    '客户': np.random.choice(['张三','李四','王五','赵六','钱七'], n),
    '城市': np.random.choice(['昆明','北京','上海','广州'], n),
    '品类': np.random.choice(['数码','服饰','食品','图书'], n),
    '单价': np.random.randint(20,500,n).astype(float),
    '数量': np.random.randint(1,6,n),
})
base.loc[5,'城市']=np.nan
base.loc[12,'城市']=np.nan
base.loc[20,'单价']=np.nan
base.loc[3,'客户']=' 张三'
base.loc[15,'客户']='李四 '
base.loc[8,'品类']=' 数码'
base['金额']=base['单价']*base['数量']
df=pd.concat([base, base.iloc[[2,10]]], ignore_index=True)  # 追加2行完全重复
print(df.shape)
df.head()
```
