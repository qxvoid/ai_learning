# Pandas 第1课：Series 与 DataFrame

> **Pandas 是什么**：建立在 NumPy 之上的**表格数据分析库**。
> NumPy 的数组没有列名、不擅长缺失值和混合类型；Pandas 专门解决这些真实数据的痛点。
>
> 两大核心对象：
> - **Series**：一维“带标签的数组”
> - **DataFrame**：二维“带行标签和列名的表”（你可以理解为 Excel 表 / SQL 表）
>
> 参考：Pandas 官方文档「10分钟上手」、清华 Pandas 讲义。

````python
import pandas as pd
import numpy as np
print(pd.__version__)
````

## 1. Series：一维带标签数组

一个 Series = **数据（values）+ 索引（index，标签）**。

```python
s = pd.Series([70, 85, 90], index=['张三','李四','王五'])
```

- 不指定 index 时，默认 0,1,2...
- 既有 NumPy 数组的运算能力，又能靠标签取值

````python
s=pd.Series([70,85,90],index=['张三','李四','王五'])
print(s)
print(s.shape)
print("values:",s.to_numpy(), type(s.values))
print("index:",s.index)
print("按标签取:",s['李四']," 按位置取:",s.iloc[1])
````

### Series 的常用操作

- 运算（逐元素，和 ndarray 一样）
- 布尔筛选
- 统计方法
- `name` 属性（它来自某列时会带列名）

````python
s=pd.Series([70,85,90,55],index=list('ABCD'))
print(s*0.5)
print(s[s>=70])
print("mean:",s.mean()," max:",s.max())
````

### 从字典创建 Series（键变成标签）

````python
d={'苹果':5,'香蕉':8,'橙子':3}
s=pd.Series(d)
print(s)
s2=pd.Series(d,index=['香蕉','苹果','葡萄'])  # 按指定标签取，缺失变NaN
print(s2)
````

## 2. DataFrame：二维表

DataFrame = 多个 Series 按列拼起来，每列共享同一套行索引。

### 创建方式1：字典（键=列名，值=列数据）

````python
df=pd.DataFrame({
    '姓名':['张三','李四','王五'],
    '语文':[70,85,90],
    '数学':[88,60,95]
})
print(df)
````

### 创建方式2：从二维 ndarray（可指定 columns/index）

````python
data=np.random.randint(50,101,(3,2))
df=pd.DataFrame(data,columns=['语文','数学'],index=['张三','李四','王五'])
print(df)
````

### 创建方式3：列表套字典（每个字典=一行，键=列名）

这是 JSON 风格数据转表的常见形式。

````python
rows=[{'姓名':'张三','语文':70,'数学':88},
      {'姓名':'李四','语文':85,'数学':60},
      {'姓名':'王五','语文':90,'数学':95,'英语':80}]
df=pd.DataFrame(rows)
print(df)   # 王五多出英语列，其他人该列是NaN
````

## 3. DataFrame 核心属性与速览方法（必会）

| 属性/方法 | 作用 |
|---|---|
| `shape` | (行数,列数) |
| `index` | 行标签 |
| `columns` | 列名 |
| `dtypes` | 每列类型 |
| `values` / `to_numpy()` | 转 ndarray |
| `head(n)/tail(n)` | 前/后 n 行（默认5） |
| `info()` | 类型+非空数量+内存（**看数据第一眼看这个**） |
| `describe()` | 数值列的统计摘要（均值/分位数等） |
| `nunique()` | 每列不同值个数 |

````python
df=pd.DataFrame(np.random.randint(40,101,(20,3)),
                columns=['语文','数学','英语'])
print(df.shape, list(df.columns))
df.info()
print(df.describe().round(1))
````

## 4. 列的访问、新增、删除

```python
df['语文']        # 取一列 -> Series
df[['语文','数学']] # 取多列 -> DataFrame
df['总分']=...     # 新增列
df['等级']=...     # 可直接赋一组值
del df['列']       # 删除列
```

````python
df=pd.DataFrame({'姓名':['张三','李四','王五'],
                 '语文':[70,85,90],'数学':[88,60,95]})
df['总分']=df['语文']+df['数学']
df['均分']=(df[['语文','数学']].mean(axis=1)).round(1)
print(df)
print(df[['姓名','均分']])
````

### 用 np.where / np.select 生成新列（条件造列）

````python
df['语文等级']=np.where(df['语文']>=80,'优','及格')
print(df[['姓名','语文','语文等级']])
conds=[df['均分']>=85, df['均分']>=60]
df['等级']=np.select(conds,['优','及格'],default='不及格')
print(df[['姓名','均分','等级']])
````

## 5. 课堂练习

1. 创建一个 DataFrame，含 20 名学生的 `学号、语文、数学、英语`（数据自定或随机）
2. 新增 `总分`、`平均分`（保留1位小数）两列
3. 用条件新增 `是否全及格`（所有科目≥60 为 是）
4. 打印 `info()` 和只含 `学号、平均分` 的表
5. 统计语文列有多少个不同分数

````python
import pandas as pd
import numpy as np
np.random.seed(42)# 建议加seed，结果可复现

df = pd.DataFrame(np.random.randint(40,101,(20,3)),
                  columns=['语文','数学','英语'])
df['学号'] = [f'学号{i}' for i in range(100001,100021)]

# 聚合永远点名列，不依赖"当前有几列"
df['总分']   = df[['语文','数学','英语']].sum(axis=1)
df['平均分'] = df[['语文','数学','英语']].mean(axis=1).round(1)
df['是否全及格'] = np.where(
    df[['语文','数学','英语']].min(axis=1) >= 60, '是', '否')

df.info()
print(df[['学号','平均分']])
print('语文不同分数个数:', df['语文'].nunique())
````
