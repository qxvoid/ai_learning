## 一、逐元素算术运算

`+ - * / // % **` 全部是**逐元素**运算（对应位置各自计算）：

````python
x = np.array([1, 4, 9, 16])
print("加:", x + 1)
print("除:", x / 2)
print("整除:", x // 2)
print("取余:", x % 3)
print("平方/幂:", x ** 2)
print("开方(函数):", np.sqrt(x))
````

> 注意：`*` 是逐元素相乘，**不是矩阵乘法**！矩阵乘法用 `@`（后面讲）。

## 二、常用数学函数（全是逐元素）

| 函数 | 作用 | 函数 | 作用 |
|------|------|------|------|
| `np.sqrt` | 平方根 | `np.abs` | 绝对值 |
| `np.exp` | e 的 x 次方 | `np.log` | 自然对数 ln |

````python
y = np.array([-1.5, 2.3, -0.7, 3.9])
print("abs:", np.abs(y))
print("round:", np.round(y))
print("floor:", np.floor(y))
print("sign:", np.sign(y))
print("exp:", np.round(np.exp(np.array([0.0, 1.0])), 2))
````

## 四、axis 参数详解（本课最容易懵的点，务必搞懂）

**axis 指定"沿着哪个方向操作"，被指定的那个维度会被"折叠/压缩"掉。**

```
a =
   列0 列1 列2 列3
行0 [ 0   1   2   3]
行1 [ 4   5   6   7]
行2 [ 8   9  10  11]
```

- **`axis=0`**：沿第 0 维（**从上往下，跨行**）推进 → 3 行折叠成 1 行 → 得到**每列**的结果，shape 从 (3,4) 变 (4,)
- **`axis=1`**：沿第 1 维（**从左往右，跨列**）推进 → 4 列折叠成 1 列 → 得到**每行**的结果，shape 从 (3,4) 变 (3,)

> 记忆：**axis=几，就消灭第几维**。
> axis=0 消灭行维度 → 剩下列的汇总；axis=1 消灭列维度 → 剩下行的汇总。

````python
print("每列之和 axis=0:", a.sum(axis=0))   # [12 15 18 21]
print("每行之和 axis=1:", a.sum(axis=1))   # [ 6 22 38]
print("每列最大 axis=0:", a.max(axis=0))   # [ 8  9 10 11]
print("每行平均 axis=1:", a.mean(axis=1).round(1))  # [1.5 5.5 9.5]
````

### keepdims=True：保留被折叠的维度（长度为1）

默认聚合后维度消失；加 `keepdims=True` 会保留长度为 1 的维度，**方便和原数组广播**：

````python
print("普通 axis=1:", a.sum(axis=1).shape)          # (3,)
k = a.sum(axis=1, keepdims=True)
print("keepdims:", k.shape)                          # (3,1)
print(k)
# [[ 6]
#  [22]
#  [38]]
print("可直接广播相减:\n", a - k)                    # (3,4)-(3,1)
````

### 累计运算（不折叠，逐位置累积）

````python
print("累计和 cumsum:")
print(a.cumsum(axis=0))   # 每列从上往下累加
# [[ 0  1  2  3]
#  [ 4  6  8 10]
#  [12 15 18 21]]
print("\n一维累计:", np.arange(1,6).cumsum())  # [ 1  3  6 10 15]
````

## 五、排序

- `np.sort(a)`：返回排好序的**副本**（不改原数组）
- `np.argsort(a)`：返回"排好序所需的索引"（用于按某列重排，你第一课用过）
- 都支持 axis

````python
s = np.array([3, 1, 4, 1, 5, 9, 2])
print("sort:", np.sort(s))
print("argsort(索引):", np.argsort(s))

m = np.array([[8, 1],[3, 6]])
print("按列排序 axis=0:\n", np.sort(m, axis=0))
print("按行排序 axis=1:\n", np.sort(m, axis=1))
````

## 六、条件与截断操作

| 函数 | 作用 |
|------|------|
| `np.where(cond, x, y)` | 满足条件取 x，否则取 y |
| `np.clip(a, lo, hi)` | 小于 lo 的变 lo，大于 hi 的变 hi（截断） |
| `np.maximum(a, b)` | 逐元素取较大值（ReLU 常用 `np.maximum(x,0)`） |
| `np.minimum(a, b)` | 逐元素取较小值 |

````python
z = np.array([-5, 120, 30, -1, 99])
print("clip 到 [0,100]:", np.clip(z, 0, 100))
print("maximum(和0取大):", np.maximum(z, 0))
print("where:", np.where(z > 50, "高", "低"))
````

## 七、去重 np.unique

返回数组中所有唯一值（自动排序）；`return_counts=True` 同时统计出现次数：

````python
u = np.array([2, 1, 2, 3, 1, 3, 3])
vals, counts = np.unique(u, return_counts=True)
print("唯一值:", vals)
print("次数:", counts)
````

## 八、拼接与分割

| 函数 | 作用 |
|------|------|
| `np.concatenate([a,b], axis=0)` | 沿指定轴拼接（最通用） |
| `np.vstack([a,b])` | 纵向拼（增加行，相当于 axis=0） |
| `np.hstack([a,b])` | 横向拼（增加列，相当于 axis=1） |
| `np.split(a, 位置列表)` | 分割 |

````python
x = np.array([[1, 2],[3, 4]])
y = np.array([[5, 6],[7, 8]])
print("纵向 vstack:\n", np.vstack([x, y]))   # 4行2列
print("横向 hstack:\n", np.hstack([x, y]))   # 2行4列
print("拼接一维:", np.concatenate([np.array([1,2]), np.array([3,4])]))
````

> 拼接要求：除了拼接方向的维度，其他维度必须一致。

## 九、重复：tile 与 repeat

- `np.tile(a, 重复布局)`：把**整个数组**像铺瓷砖一样重复
- `np.repeat(a, n, axis)`：把**每个元素**逐个重复

````python
a = np.array([[1, 2],[3, 4]])
print("tile 整体铺(2行1列次):\n", np.tile(a, (2, 1)))
print("repeat 逐元素(每元素2次,沿列):\n", np.repeat(a, 2, axis=1))
````

## 十、线性代数（np.linalg）

### 矩阵乘法 `@`（或 np.dot）
- `(m,k) @ (k,n) -> (m,n)`，**中间维度必须相等**
- 一维向量间 `@` = 内积（点积）

````python
A = np.array([[1, 2],[3, 4]])   # (2,2)
B = np.array([[5, 6],[7, 8]])   # (2,2)
print("逐元素乘 A*B:\n", A*B)
print("矩阵乘 A@B:\n", A@B)

v = np.array([1, 2])
print("向量内积 v@v:", v@v)   # 5
````

### linalg 常用函数

| 函数 | 作用 |
|------|------|
| `np.linalg.inv(A)` | 逆矩阵（方阵） |
| `np.linalg.det(A)` | 行列式 |
| `np.linalg.eig(A)` | 特征值与特征向量 |
| `np.linalg.solve(A,b)` | 解线性方程组 Ax=b（推荐，比手动求逆稳） |
| `np.linalg.norm(x)` | 范数（向量长度） |

````python
A = np.array([[2.0, 1.0],[1.0, 3.0]])
print("行列式:", np.linalg.det(A).round(2))
print("逆矩阵:\n", np.linalg.inv(A).round(2))

b = np.array([4.0, 5.0])
x = np.linalg.solve(A, b)     # 解 Ax=b
print("方程组的解:", x.round(2))
print("验证 A@x:", (A@x).round(2))
````
