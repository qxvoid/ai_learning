## 1. 从 Python 列表创建数组

- 一维数组：传入普通列表
- 二维数组：传入嵌套列表（理解为「矩阵」，外层是行）

```python
# 一维数组
arr1 = np.array([1, 2, 3, 4, 5])
print("一维数组:", arr1)

# 二维数组（2 行 3 列）
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])
print("二维数组:\n", arr2)
```

## 2. 生成固定规则的数组

| 函数 | 作用 |
|------|------|
| `np.zeros(shape)` | 全 0 数组 |
| `np.ones(shape)` | 全 1 数组 |
| `np.full(shape, value)` | 全指定值数组 |
| `np.eye(n)` | n 阶单位矩阵 |

```python
print("全 0 数组（3行4列）:\n", np.zeros((3, 4)))
print("\n全 1 数组（2行3列）:\n", np.ones((2, 3)))
print("\n全为 666 的数组:\n", np.full((2, 2), 666))
print("\n3 阶单位矩阵:\n", np.eye(3))
```

## 3. 生成序列数组

- `np.arange(start, stop, step)`：类似 range，**步长可以是小数**
- `np.linspace(start, stop, num)`：生成 num 个**等距**元素（包含终点）

```python
print("arange（0到10，步长2）:", np.arange(0, 10, 2))
print("linspace（0到10，均分5份）:", np.linspace(0, 10, 5))
```

## 4. 生成随机数组

- `np.random.rand(shape)`：[0, 1) 均匀分布
- `np.random.randint(low, high, size)`：随机整数（左闭右开）
- `np.random.randn(shape)`：标准正态分布（均值 0，方差 1）

```python
print("[0,1) 均匀分布:\n", np.random.rand(2, 3))
print("\n0-99 随机整数:\n", np.random.randint(0, 100, size=(3, 3)))
print("\n标准正态分布:\n", np.random.randn(3, 2))
```

## 5. 数组的四大核心属性

以二维数组 `arr2` 为例：

| 属性 | 含义 |
|------|------|
| `.ndim` | 维度（轴的个数） |
| `.shape` | 形状（每个维度的长度，如行数、列数） |
| `.size` | 元素总个数 |
| `.dtype` | 元素的数据类型 |

```python
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])
print("维度 ndim:", arr2.ndim)
print("形状 shape:", arr2.shape)
print("元素总数 size:", arr2.size)
print("元素类型 dtype:", arr2.dtype)
```
