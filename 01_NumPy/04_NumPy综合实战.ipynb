## 题 4：判断能否广播；能则写结果 shape，不能说明原因

- (a) `(3,4)` 和 `(4,)`
- (b) `(3,4)` 和 `(3,)`
- (c) `(3,4)` 和 `(3,1)`
- (d) `(2,3,4)` 和 `(2,4)`
- (e) `(4,1)` 和 `(3,)`


(a) 右对齐，左补1拉伸，shape=(3,4)
(b) 右边无法对其，无法广播
(c) 左对齐，右边拉伸即可，shape=(3,4)
(d) 中间加一维，[:,np.newaxis,:],中间拉伸。shape=(2,4,3)
(e) 先右边对其，(4,1)先拉伸，(3,)左补1，拉伸，shape=(4,3)

---
**【批改订正】**
- (c) 准确说法：右对齐后，第0维 3=3 相等、第1维 1 拉伸成 4
- (d) `(2,3,4)+(2,4)`：短数组补成 `(1,2,4)`，中间维 **3 vs 2 冲突 → 自动广播报错**。自动补维只补最左边；手动插中间 `b[:,None,:]` 变成 `(2,1,4)` 才能广播，结果是 `(2,3,4)`（不是 (2,4,3)）。

## 题 9：线性代数

1. 给定 `A=[[2,1],[1,3]]`、`b=[4,5]`，用 `solve` 解 Ax=b，并验证 A@x≈b
2. 求 A 的行列式和逆矩阵，验证 A @ inv(A) ≈ 单位矩阵
3. 自行选定 M、K、N，构造 `(M,K)` 和 `(K,N)` 两个矩阵做乘法，验证结果 shape

```python
import numpy as np

# 1. 解方程组 Ax=b
A = np.array([[2.,1],[1,3]])
b = np.array([4.,5])
x = np.linalg.solve(A,b)
print("解 x =", x)
print("验证 A@x =", A @ x)

# 2. 行列式与逆矩阵
print("行列式 det =", np.linalg.det(A))
invA = np.linalg.inv(A)
print("逆矩阵:\n", invA)
print("A@invA:\n", (A@invA).round(6))

# 3. 矩阵乘法 shape：(M,K)@(K,N) -> (M,N)
M, K, N = 2, 3, 4
X = np.random.randint(0,10,(M,K))
Y = np.random.randint(0,10,(K,N))
print("X@Y shape =", (X@Y).shape)
```

## 项目一：学生成绩分析

**数据**：10 名学生、5 门课程，用 `seed=42` 生成 50~100 的随机整数成绩。

**需求（按顺序实现，每步打印结果）**
1. 计算每名学生的总分、平均分（平均分保留 1 位小数）
2. 按平均分从高到低排名，输出排名后的学生编号顺序及对应平均分
3. 评定每人等级：≥90 优、≥80 良、≥60 及格、其余不及格
4. 找出「所有科目都 ≥80」的全优生编号
5. 找出「有任意一门 <60」的学生编号
6. 统计每门课的：平均分、最高分、及格率（≥60 的比例）

```python
import numpy as np
np.random.seed(42)
a = np.random.randint(50,101,size=(10,5))
print(a)
ans1_1 = a.sum(axis=1)
ans1_2 = a.mean(axis=1).round(1)

idx2 = np.argsort(a.mean(axis=1))[::-1]
temp = ans1_2[idx2]
idx2 += 1
ans2 = np.column_stack([idx2, temp])
print(ans2)

level = np.where(temp >= 90,'优',
        np.where(temp >= 80,'良',
        np.where(temp >= 60,'及格','不及格')))
ans3 = np.column_stack([idx2, temp, level])
print(ans3)
temp3 = np.min(a,axis=1)
ans4 = []
for i in range(10):
    if temp3[i] >= 80:
        ans4.append(i+1)
ans4 = np.array(ans4)
print("全优生编号:", ans4)
ans5 = []
for i in range(10):
    if temp3[i] < 60:
        ans5.append(i+1)
ans5 = np.array(ans5)
print("有任意一门不及格的编号:", ans5)
m = np.mean(a,axis=0).round(1)
n = np.max(a,axis=0)
k = (a>=60).mean(axis=0).round(1)
print("每门课平均分:", m)
print("每门课最高分:", n)
print("每门课及格率:", k)
```
